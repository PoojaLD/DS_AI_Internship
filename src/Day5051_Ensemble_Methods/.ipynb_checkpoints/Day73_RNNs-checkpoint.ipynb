Exercise Solutions

1. If an RNN is used to predict the next character in a sequence, what should be the dimensionality of the output layer?
The dimensionality of the output layer should be equal to the size of the vocabulary (the total number of unique characters in the dataset). Each dimension typically represents the probability of a specific character being the next in the sequence.

2. Why are RNNs capable of modeling dependencies across long sequences?
They are capable because of their recurrent structure, where the hidden state from the previous time step is passed as an input to the current time step. This creates a "memory" that allows information from earlier parts of the sequence to persist and influence later predictions.

3. What challenges arise when backpropagating through long sequences?
The primary challenges are the vanishing gradient and exploding gradient problems.Vanishing: Gradients become extremely small as they are multiplied back through many time steps, making it difficult for the network to learn long-term dependencies.Exploding: Gradients become excessively large, causing the model's weights to update too drastically and making training unstable.

4. What are the limitations of basic RNN-based language models?
Short-term Memory: Due to vanishing gradients, basic RNNs struggle to retain information over very long gaps.Sequential Processing: They process data one step at a time, which prevents parallelization and makes training slower compared to modern architectures like Transformers.Fixed-length Context: They can struggle to prioritize relevant historical information over "noise" in very long sequences.
